```python

import torch
import torch.nn.functional as F

# Input
text = "Artificial Intelligence is powerful"
inputs = tokenizer(text, return_tensors="pt")

# Get first Transformer block
attention = model.transformer.h[0].attn

# Token + Positional Embeddings
input_ids = inputs["input_ids"]
position_ids = torch.arange(input_ids.shape[1]).unsqueeze(0)

hidden = (
    model.transformer.wte(input_ids)
    + model.transformer.wpe(position_ids)
)

# QKV Projection
qkv = attention.c_attn(hidden)

# Split Q, K, V
q, k, v = qkv.split(768, dim=2)

# Reshape into attention heads
batch, seq_len, _ = q.shape
heads = 12
head_dim = 64

q = q.view(batch, seq_len, heads, head_dim).transpose(1, 2)
k = k.view(batch, seq_len, heads, head_dim).transpose(1, 2)
v = v.view(batch, seq_len, heads, head_dim).transpose(1, 2)

print("Q:", q.shape)
print("K:", k.shape)
print("V:", v.shape)
```

```python
import torch
import torch.nn.functional as F

# 1. Input
text = "Artificial Intelligence is powerful"
inputs = tokenizer(text, return_tensors="pt")
input_ids = inputs["input_ids"]

# 2. Position IDs
position_ids = torch.arange(input_ids.shape[1]).unsqueeze(0)

# 3. GPT-2 embeddings + LayerNorm
hidden = model.transformer.wte(input_ids) + model.transformer.wpe(position_ids)
block = model.transformer.h[0]
hidden = block.ln_1(hidden)

# 4. Generate Q, K, V
qkv = block.attn.c_attn(hidden)
q, k, v = qkv.split(768, dim=2)

# 5. Split into 12 attention heads
batch_size, seq_len, _ = q.shape
heads = 12
head_dim = 64

q = q.view(batch_size, seq_len, heads, head_dim).transpose(1, 2)
k = k.view(batch_size, seq_len, heads, head_dim).transpose(1, 2)
v = v.view(batch_size, seq_len, heads, head_dim).transpose(1, 2)

# 6. Select first attention head
q_head = q[:, 0, :, :]
k_head = k[:, 0, :, :]
v_head = v[:, 0, :, :]

# 7. Calculate attention scores
scores = torch.matmul(q_head, k_head.transpose(-2, -1))
scores = scores / (head_dim ** 0.5)

# 8. Causal mask
causal_mask = torch.triu(
    torch.ones(seq_len, seq_len, dtype=torch.bool),
    diagonal=1
)
scores = scores.masked_fill(causal_mask, float("-inf"))

# 9. Softmax
attention_weights = F.softmax(scores, dim=-1)

# 10. Context vector
context = torch.matmul(attention_weights, v_head)

print("Q Shape:", q_head.shape)
print("Attention Weights Shape:", attention_weights.shape)
print("Context Vector Shape:", context.shape)
```

```python
# PDF Loading & Text Extraction
# Load PDF

loader = PyPDFLoader(pdf_path)

# Extract text from PDF
pages = loader.load()

# Check extracted data
print("Total Pages:", len(pages))

print("\nFirst Page Content:\n")
print(pages[0].page_content)
```

```python
# Split PDF into Chunks

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = text_splitter.split_documents(pages)

print("Total Chunks:", len(chunks))

print("\nFirst Chunk:\n")
print(chunks[0].page_content)
```

```python
# Generate Embeddings

# Load embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Extract text from all chunks
chunk_texts = [chunk.page_content for chunk in chunks]

# Generate embeddings
embeddings = embedding_model.encode(
    chunk_texts,
    convert_to_tensor=True
)

print("Total Chunks:", len(chunk_texts))
print("Embeddings Shape:", embeddings.shape)
print("\nFirst Chunk Embedding:")
print(embeddings[0])
```

```python
# Import Integration

from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

print("Integration Ready!")
```

```python
# Store Documents in ChromaDB

embedding_function = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_function,
    collection_name="iit_aiml_documents"
)

print("Total Stored Documents:", vectorstore._collection.count())
```

```python
#Similarity Search
#Query Search

query = "What is Artificial Intelligence?"

results = vectorstore.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results):
    print(f"\nResult {i+1}:")
    print(doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
# The 2nd Retriever Pipeline
# Retriever Creation
# Create Retriever

retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

# Retrieve relevant documents
retrieved_docs = retriever.invoke(
    "what is Manual Feature Engineering?"
)

print("Retrieved Documents:", len(retrieved_docs))

for i, doc in enumerate(retrieved_docs):
    print(f"\nDocument {i+1}:")
    print(doc.page_content)
```

```python
# final prompt how look
# this code for testing/inspection

formatted_prompt = prompt.invoke({
    "context": context,
    "question": question
})

print(formatted_prompt.to_string())
```

```python
# Connect RAG with GPT-2

# Convert formatted prompt into text
input_text = formatted_prompt.to_string()

# Tokenize the prompt
inputs = tokenizer(input_text, return_tensors="pt")

# Generate answer using GPT-2
outputs = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_new_tokens=150,
    do_sample=True,
    temperature=0.7,
    pad_token_id=tokenizer.eos_token_id
)

# Extract only newly generated text
new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

answer = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print("RAG Answer:")
print(answer)
```

```python
# LoRA Configuration

from peft import LoraConfig, get_peft_model, TaskType

lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=["c_attn"]
)

model = get_peft_model(model, lora_config)

model.print_trainable_parameters()
```

```python
# Create Training Dataset

training_data = [
    {
        "instruction": "What is Artificial Intelligence?",
        "response": "Artificial Intelligence is a field of computer science that enables machines to perform tasks that normally require human intelligence."
    },
    {
        "instruction": "What is Machine Learning?",
        "response": "Machine Learning is a subset of Artificial Intelligence that allows computers to learn patterns from data and make predictions."
    },
    {
        "instruction": "What is Deep Learning?",
        "response": "Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers to learn complex patterns."
    },
    {
        "instruction": "What is Natural Language Processing?",
        "response": "Natural Language Processing is a field of AI that enables computers to understand, process, and generate human language."
    },
    {
        "instruction": "What is Generative AI?",
        "response": "Generative AI is a type of artificial intelligence that can generate new content such as text, images, audio, and code."
    },
    {
        "instruction": "What is a Neural Network?",
        "response": "A Neural Network is a machine learning model inspired by the human brain that learns patterns using interconnected artificial neurons."
    },
    {
        "instruction": "What is Computer Vision?",
        "response": "Computer Vision is a field of AI that enables computers to interpret and understand visual information such as images and videos."
    },
    {
        "instruction": "What is an LLM?",
        "response": "A Large Language Model is an AI model trained on large amounts of text data to understand and generate human language."
    }
]

print("Total Training Examples:", len(training_data))

for item in training_data[:2]:
    print("\nInstruction:", item["instruction"])
    print("Response:", item["response"])
```

```python
# Dataset Formatting & Tokenization

from datasets import Dataset

# Convert instruction-response pairs into text format
formatted_data = []

for item in training_data:
    text = (
        f"### Instruction:\n{item['instruction']}\n\n"
        f"### Response:\n{item['response']}"
    )

    formatted_data.append({"text": text})

# Convert into Hugging Face Dataset
dataset = Dataset.from_list(formatted_data)

print("Dataset:", dataset)
print("\nFirst Training Example:\n")
print(dataset[0]["text"])
```

```python
# Tokenize Training Dataset

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=256
    )

tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True
)

print("Tokenization Completed!")
print("\nTokenized Dataset:")
print(tokenized_dataset)

print("\nInput IDs:")
print(tokenized_dataset[0]["input_ids"][:30])

print("\nDecoded Text:")
print(tokenizer.decode(tokenized_dataset[0]["input_ids"]))
```

```python
# Data Collator

from transformers import DataCollatorForLanguageModeling

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

print("Data Collator Ready!")
```

```python
# Training Configuration

from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./gpt2-lora-finetuned",
    per_device_train_batch_size=2,
    num_train_epochs=10,
    learning_rate=2e-4,
    logging_steps=1,
    save_strategy="epoch",
    fp16=True,
    report_to="none"
)

print("Training Configuration Ready!")
```

```python
# Train the GPT-2 Model using LoRA

from transformers import Trainer

model.config.use_cache = False

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator
)

training_result = trainer.train()

print("\nFine-tuning Completed!")
print(training_result)
```

```python
# Save Fine-tuned LoRA Adapter

save_path = "./gpt2-lora-finetuned"

model.save_pretrained(save_path)
tokenizer.save_pretrained(save_path)

print("Fine-tuned LoRA Adapter Saved Successfully!")
print("Location:", save_path)
```

```python
# Fine-tuned Model Testing

from peft import PeftModel

base_model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2")

fine_tuned_model = PeftModel.from_pretrained(
    base_model,
    "./gpt2-lora-finetuned"
)

fine_tuned_model.eval()

print("Fine-tuned Model Loaded Successfully!")
```

```python
# Compare Both Models

# Original GPT-2 model
original_model = AutoModelForCausalLM.from_pretrained(
    "openai-community/gpt2"
)

original_model.eval()

# Test question
question = "What is Machine Learning?"

prompt_text = (
    f"### Instruction:\n{question}\n\n"
    f"### Response:\n"
)

inputs = tokenizer(prompt_text, return_tensors="pt").to("cuda")

# Original model response
with torch.no_grad():
    original_output = original_model.generate(
        **inputs,
        max_new_tokens=60,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

# Fine-tuned model response
with torch.no_grad():
    finetuned_output = fine_tuned_model.generate(
        **inputs,
        max_new_tokens=60,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

# Decode generated responses
original_answer = tokenizer.decode(
    original_output[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

finetuned_answer = tokenizer.decode(
    finetuned_output[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print("QUESTION:", question)

print("\n--- ORIGINAL GPT-2 ---")
print(original_answer)

print("\n--- FINE-TUNED GPT-2 ---")
print(finetuned_answer)
```
